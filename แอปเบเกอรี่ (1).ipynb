```python
from IPython.display import HTML, display

display(HTML("""
<!DOCTYPE html>
<html lang="th">
<head>
<meta charset="UTF-8">
<title>Bakery App</title>

<style>
    body {
        font-family: Arial, sans-serif;
        background: #fff3e6;
        margin: 0;
        padding: 20px;
    }

    .app {
        max-width: 500px;
        margin: auto;
        background: white;
        border-radius: 20px;
        padding: 25px;
        box-shadow: 0 5px 20px rgba(0,0,0,0.15);
    }

    h1 {
        text-align: center;
        color: #8b4513;
    }

    .menu {
        display: flex;
        gap: 10px;
        margin-top: 20px;
        justify-content: center;
        flex-wrap: wrap;
    }

    button {
        background: #d2691e;
        color: white;
        border: none;
        padding: 10px 15px;
        border-radius: 10px;
        cursor: pointer;
    }

    button:hover {
        background: #a0522d;
    }

    .product {
        background: #fff8f0;
        border: 1px solid #f0c9a4;
        border-radius: 15px;
        padding: 15px;
        margin: 10px 0;
    }

    .product h3 {
        color: #8b4513;
        margin: 5px 0;
    }

    input {
        width: 60px;
        padding: 7px;
        border-radius: 8px;
        border: 1px solid #ccc;
    }

    #total {
        font-size: 20px;
        font-weight: bold;
        color: #d2691e;
        margin-top: 20px;
        text-align: center;
    }

    #status {
        font-size: 18px;
        font-weight: bold;
        color: green;
        text-align: center;
        margin-top: 10px;
    }
</style>
</head>

<body>

<div class="app">

    <h1>🍰 Bakery App 🧁</h1>

    <div id="products">

        <div class="product">
            <h3>🍰 เค้กช็อกโกแลต</h3>
            <p>ราคา 50 บาท</p>
            จำนวน:
            <input type="number" id="cake" value="0" min="0">
        </div>

        <div class="product">
            <h3>🍪 คุกกี้</h3>
            <p>ราคา 20 บาท</p>
            จำนวน:
            <input type="number" id="cookie" value="0" min="0">
        </div>

        <div class="product">
            <h3>🥐 ครัวซองต์</h3>
            <p>ราคา 35 บาท</p>
            จำนวน:
            <input type="number" id="croissant" value="0" min="0">
        </div>

        <div class="product">
            <h3>🍞 ขนมปัง</h3>
            <p>ราคา 25 บาท</p>
            จำนวน:
            <input type="number" id="bread" value="0" min="0">
        </div>

    </div>

    <!-- ปุ่มทั้งหมดอยู่ด้านล่าง -->
    <div class="menu">
        <button onclick="calculateTotal()">คำนวณราคา</button>
        <button onclick="clearOrder()">ล้างรายการ</button>
        <button onclick="order()">🛒 สั่งซื้อ</button>
    </div>

    <div id="total">ยอดรวม: 0 บาท</div>

    <div id="status"></div>

</div>

<script>

// ฟังก์ชันคำนวณราคา
function calculateTotal() {

    let cake = Number(document.getElementById("cake").value);
    let cookie = Number(document.getElementById("cookie").value);
    let croissant = Number(document.getElementById("croissant").value);
    let bread = Number(document.getElementById("bread").value);

    let total =
        (cake * 50) +
        (cookie * 20) +
        (croissant * 35) +
        (bread * 25);

    document.getElementById("total").innerHTML =
        "ยอดรวม: " + total + " บาท";

    // ตอนคำนวณราคา จะไม่แสดงข้อความสั่งซื้อเสร็จสิ้น
    document.getElementById("status").innerHTML = "";
}


// ฟังก์ชันสั่งซื้อ
function order() {

    // คำนวณราคาก่อน
    calculateTotal();

    // แสดงข้อความสั่งซื้อเสร็จสิ้น
    document.getElementById("status").innerHTML =
        "✅ สั่งซื้อเสร็จสิ้น";
}


// ฟังก์ชันล้างรายการ
function clearOrder() {

    document.getElementById("cake").value = 0;
    document.getElementById("cookie").value = 0;
    document.getElementById("croissant").value = 0;
    document.getElementById("bread").value = 0;

    document.getElementById("total").innerHTML =
        "ยอดรวม: 0 บาท";

    document.getElementById("status").innerHTML = "";
}

</script>

</body>
</html>
"""))
```
